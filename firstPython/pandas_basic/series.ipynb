```python
import pandas as pd
```

```python
# * 리스트로 Series 생성 -> 인덱스 안 주면 기본값 0,1,2... 자동 부여
data = [10, 20, 30, 40, 50]
series = pd.Series(data)

print(series)
```

```python
# * index 파라미터로 인덱스(라벨) 직접 지정 -> one dimensional "labeled" array
data = [10, 20, 30, 40, 50]
index_labels = ['A', 'B', 'C', 'D', 'E']
series = pd.Series(data, index=index_labels)

print(series)
```

```python
# * 맨 앞 n개 값 (DataFrame에서도 똑같이 쓸 수 있음)
series.head(1)
```

```python
# * 맨 뒤 n개 값
series.tail(2)
```

```python
# * 숫자형 데이터 통계 요약 (개수/평균/표준편차/최솟값/사분위수/최댓값)
series.describe()
```

```python
# * 전체 값의 합
series.sum()
```

```python
# * 값별로 몇 번씩 등장했는지 개수 세기
series.value_counts()
```

```python
# * 딕셔너리로 Series 만들 준비 -> key가 인덱스가 될 예정
city_population_data = {"tokyo": 37000000, "seoul": 9700000, "beijing": 21000000}
```

```python
# * 딕셔너리로 Series 생성 -> key가 자동으로 인덱스(라벨)가 됨, value가 값이 됨
pd.Series(city_population_data)
```

```python
# * 인덱스가 다른 Series끼리 연산하면 어떻게 되는지 확인하기 위한 준비
s1 = {"a": 1, "b": 2}
s2 = {"a": 10, "b": 5, "c": 5}
```

```python
# * Series끼리 연산은 "같은 인덱스"끼리 자동으로 짝지어서 계산됨 (index alignment)
# * 한쪽에만 있는 인덱스(c)는 상대쪽 값이 없어서 결과가 NaN
pd.Series(s1) + pd.Series(s2)
```
