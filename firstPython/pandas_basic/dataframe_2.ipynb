```python
# * pandas 불러오기
import pandas as pd
```

```python
# * 정렬(sort_values) 연습용 DataFrame 생성
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'San Francisco', 'Los Angeles']
}

df = pd.DataFrame(data)
```

```python
# * sort_values -> 특정 열(by) 기준으로 행 정렬
# * ascending=True(기본값) -> 오름차순(작은 값 -> 큰 값), ascending=False -> 내림차순(큰 값 -> 작은 값)
df.sort_values(by='Age', ascending=False)
```

```python
# * Age 열의 평균
df['Age'].mean()
```

```python
# * 열 이름 바꾸기
data = {
    'A': [1, 2, 3],
    'B': [4, 5, 6],
    'C': [7, 8, 9]
}

df = pd.DataFrame(data)

# * rename(columns={기존이름: 새이름}) -> inplace=True로 원본 자체 수정
df.rename(columns={'A': 'Column_A', 'B': 'Column_B', 'C': 'Column_C'}, inplace=True)
print(df)
```

```python
# * DataFrame 이어붙이기(concat) 연습
data1 = {
    'Name': ['Alice', 'Bob'],
    'Age': [25, 30]
}

data2 = {
    'Name': ['Charlie', 'David'],
    'Age': [35, 40]
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

# * pd.concat([df1, df2]) -> 세로로 이어붙임, 인덱스는 각자 것 그대로 유지 (0,1,0,1)
df_concatenated = pd.concat([df1, df2])
print(df_concatenated)

print("\n")

# * ignore_index=True -> 기존 인덱스 무시하고 0부터 새로 번호 매김
df_concatenated = pd.concat([df1, df2], ignore_index=True)
print(df_concatenated)
```

```python
# * DataFrame 합치기(merge/join) 연습
data1 = {
    'ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'Sarah']
}

data2 = {
    'ID': [1, 2, 3],
    'Age': [25, 30, 35]
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

# * merge -> 공통 열(on='ID') 기준으로 합침, how='inner'는 양쪽 다 있는 ID만 남김 (SQL JOIN이랑 비슷)
# * how는 'outer'(양쪽 다 포함, 없는 값은 NaN), 'right'(오른쪽 기준) 등도 가능
df_merged = pd.merge(df1, df2, how='inner', on='ID')
print(df_merged)

print("\n")

# * join은 "인덱스" 기준으로만 합칠 수 있어서, 그냥 열이었던 ID를 먼저 인덱스로 바꿔야 함
# * set_index('ID') -> ID 열을 뽑아서 인덱스(행 이름표)로 승격시킴 (더 이상 일반 열이 아님)
df1.set_index('ID', inplace=True)
df2.set_index('ID', inplace=True)

# * join -> 인덱스 기준으로 합침 (merge랑 다르게 열이 아니라 index를 기준으로 씀)
joined_df = df1.join(df2)
print(joined_df)
```

```python
# * groupby + 여러 집계 함수 연습
data = {
    'Category': ['A', 'A', 'B', 'B', 'A', 'B'],
    'Value': [10, 20, 30, 40, 50, 60]
}

df = pd.DataFrame(data)

# * Category별로 그룹 지어놓기 (아직 계산은 안 함, 그룹 객체만 만듦)
group_by_category = df.groupby('Category')

# * 그룹별 평균
print(group_by_category.mean())

# * 그룹별 표준편차
print(group_by_category.std())

# * 그룹별 통계 요약 한 번에 (count/mean/std/min/25%/50%/75%/max)
print(group_by_category.describe())

# * agg({열이름: 함수이름}) -> 딕셔너리로 "이 열엔 이 함수" 매핑을 지정
# * {'Value': 'sum'} -> key='Value'(적용할 열), value='sum'(적용할 함수, 문자열로)
# * 지금은 열이 Value 하나뿐이라 .sum()이랑 결과가 똑같아 보이지만,
# * agg({'A':'sum', 'B':'mean'})처럼 열마다 다른 함수를 각각 지정할 수 있는 게 진짜 쓸모
grouped_df = group_by_category.agg({'Value': 'sum'})
print(grouped_df)
```

```python
# * MultiIndex -> 인덱스를 여러 단계(레벨)로 만들기
import pandas as pd

# * (바깥 인덱스 값, 안쪽 인덱스 값) 쌍
index_tuples = [
    ('A', 'one'),
    ('A', 'two'),
    ('B', 'one'),
    ('B', 'two'),
    ('C', 'one'),
    ('C', 'two')
]

# * names -> 각 인덱스 레벨에 붙이는 이름표 (열 이름 붙이는 거랑 같은 개념, 대상만 인덱스)
multi_index = pd.MultiIndex.from_tuples(index_tuples, names=['Letter', 'Number'])

data = {'Value': [1, 2, 3, 4, 5, 6]}
df = pd.DataFrame(data, index=multi_index)

print(df)

```

```python
# * MultiIndex 각 단계(레벨)의 이름 확인
df.index.names
```

```python
# * loc를 연달아 써서 바깥 인덱스 -> 안쪽 인덱스 순서로 접근
df.loc['A'].loc['one']
```

```python
# * MultiIndex DataFrame에도 새 열 추가는 똑같이 동작
df['Value2'] = [9,8,7,6,5,4]
```

```python
# * loc 연달아 쓰고 마지막에 열 이름까지 -> 값 하나 콕 집기
df.loc['A'].loc['one']['Value2']
```

```python
# * level -> 위에서 names=['Letter', 'Number']로 설정한 그 인덱스 이름 중 몇 번째에서 찾을지 지정
# * level=1 -> 'Number'라는 이름의 레벨에서 'one' 찾기
df.xs('one', level=1)
```
