```python
import numpy as np
import pandas as pd
```

```python
# * np.nan을 일부러 섞어서 결측치(빠진 값) 연습용 DataFrame 생성
data = {
    'A': [1, 2, np.nan, 4],
    'B': [5, np.nan, np.nan, 8],
    'C': [9, 10, 11, 12]
}

df = pd.DataFrame(data)
print(df)
```

```python
# * isna() -> 값이 NaN(결측치)인 자리마다 True/False로 표시
print(df.isna())
```

```python
# * dropna() -> NaN이 하나라도 있는 "행"을 통째로 삭제 (기본값)
df_dropped_rows = df.dropna()
```

```python
df_dropped_rows
```

```python
# * thresh=2 -> "정상 값(NaN 아닌 값)이 최소 2개는 있어야" 그 행을 살려둠
# * 무조건 NaN 있으면 삭제(dropna())보다 덜 엄격하게 걸러내고 싶을 때 씀
df_dropped_rows = df.dropna(thresh=2)
```

```python
df_dropped_rows
```

```python
# * dropna(axis=1) -> NaN이 하나라도 있는 "열"을 통째로 삭제 (axis=1이라 방향이 열)
df_dropped_columns = df.dropna(axis=1)
```

```python
df_dropped_columns
```

```python
# * fillna(값) -> NaN 자리를 지정한 값으로 채우기 (여기선 전부 0으로)
df.fillna(0)
```

```python
# * fillna(df.mean()) -> NaN을 "그 열의 평균값"으로 채우기 (열마다 다른 평균값이 들어감)
df.fillna(df.mean())
```

```python
# ! fillna(method='ffill')는 현재 pandas 버전(3.x)에서 삭제됨 -> TypeError 남
# * 지금은 df.ffill()을 직접 써야 함
# * ffill(forward fill) -> NaN을 "바로 위(이전) 값"으로 채우기
df.ffill()
```

```python
# ! fillna(method='bfill')도 마찬가지로 삭제됨 -> df.bfill()을 직접 써야 함
# * bfill(backward fill) -> NaN을 "바로 아래(다음) 값"으로 채우기
df.bfill()
```

```python
# * unique() -> 중복 제거한 값 목록 (NaN도 값으로 포함됨)
df['A'].unique()
```

```python
# * nunique() -> 중복 없는 값이 "몇 개"인지 개수만 (NaN은 기본적으로 개수에서 제외됨)
df['A'].nunique()
```

```python
# * value_counts() -> 값마다 몇 번씩 등장했는지 개수 세기
df['A'].value_counts()
```

```python
# * apply/axis 연습용 DataFrame, 인덱스를 학생 이름으로 지정
data = {
    'Math': [85, 90, 78, 92, 88],
    'Physics': [80, 85, 86, 92, 95],
    'Chemistry': [70, 75, 72, 89, 85]
}

df = pd.DataFrame(data, index=['Alice', 'Bob', 'Charlie', 'David', 'Eva'])
print(df)
```

```python
# * apply(함수, axis) -> 행 또는 열 단위로 원하는 함수를 직접 적용 (내장 함수로 안 되는 계산 할 때 씀)
# * lambda row: row.mean() -> "row라는 값을 받아서 row.mean()을 계산해라"는 뜻의 이름 없는 함수(익명 함수)
# *   def my_func(row): return row.mean() 이랑 완전히 같은 함수를 한 줄로 줄여 쓴 것
# * axis=1 -> 함수를 "행" 단위로 적용 (학생 한 명씩 -> 그 학생의 세 과목 평균)
row_average = df.apply(lambda row: row.mean(), axis=1)
```

```python
row_average
```

```python
# * axis=0 -> 함수를 "열" 단위로 적용 (과목 하나씩 -> 그 과목의 전체 학생 평균)
col_average = df.apply(lambda col: col.mean(), axis=0)
```

```python
col_average
```

```python
# * pivot_table 연습용 DataFrame 준비
data = {
    'Region': ['North', 'North', 'South', 'South', 'East', 'East', 'West', 'West'],
    'Product': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Sales': [100, 150, 200, 250, 300, 350, 400, 450]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# * pivot_table -> 엑셀 피벗 테이블이랑 같은 개념, 데이터를 표 형태로 재구성해서 요약
# * index='Region' -> 결과의 행 기준, columns='Product' -> 결과의 열 기준
# * values='Sales' -> 채울 값, aggfunc='sum' -> Region+Product 조합마다 Sales를 합산해서 채움
pivot_table = df.pivot_table(
                values='Sales',
                index='Region',
                columns='Product',
                aggfunc='sum')
```

```python
pivot_table
```
