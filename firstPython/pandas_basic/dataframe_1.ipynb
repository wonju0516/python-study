```python
# * pandas 불러오기
import pandas as pd
```

```python
import numpy as np

# * np.random.randn(행, 열) -> 정규분포(평균 0)에서 뽑은 랜덤 값으로 채워진 배열 생성
# * 4행 3열 랜덤 값 DataFrame 생성, 행 인덱스는 a~d, 열 이름은 x,y,z
df = pd.DataFrame(np.random.randn(4, 3), index=['a', 'b', 'c', 'd'], columns=['x', 'y', 'z'])
```

```python
df
```

```python
# * 열 하나만 뽑으면 타입은 Series
type(df['x'])
```

```python
# * DataFrame 자체 타입 확인
type(df)
```

```python
# * 'x' 열만 Series로 가져오기
df['x']
```

```python
# * 여러 열을 가져올 땐 리스트로 감싸기 -> 결과는 DataFrame
df[['x', 'y']]
```

```python
# * 새 열 'w' 추가
df['w'] = np.random.randn(4)
```

```python
df
```

```python
# * axis=1 -> 열 삭제, 기본은 원본이 안 바뀜(not in place) -> 삭제된 결과만 반환
df.drop('w', axis=1)
```

```python
# * 위에서 inplace 안 줬으니 w열 그대로 남아있음 (원본 안 바뀜 확인)
df
```

```python
# * inplace=True -> 원본 df 자체를 바로 수정
df.drop('w', axis=1, inplace=True)
```

```python
# * inplace=True 했으니 w열 실제로 삭제됨
df
```

```python
# * axis=0 -> 행 삭제 (기본 not in place)
df.drop('d', axis=0)
```

```python
# * 위에서 inplace 안 줬으니 d행 그대로 남아있음
df
```

```python
# * shape (4,3) -> axis=0은 첫 번째 자리(행), axis=1은 두 번째 자리(열)를 가리킴
df.shape
```

```python
# * loc -> 인덱스 '이름'으로 행 접근
df.loc['a']
```

```python
# * iloc -> 정수 '위치'로 행 접근 (0부터 시작)
df.iloc[0]
```

```python
# * loc[행이름, 열이름] -> 값 하나 콕 집어서 접근
df.loc['a', 'x']
```

```python
# * 행/열을 여러 개 리스트로 지정 -> 부분 표(서브셋) 가져오기
df.loc[['a', 'b'], ['x', 'y']]
```

```python
# * loc 슬라이싱(:) -> 라벨 기준, 끝 라벨도 "포함"됨 (일반 파이썬 슬라이싱과 다름)
df.loc['a':'c']
```

```python
# * iloc 슬라이싱(:) -> 위치 기준, 끝 위치는 "미포함" (파이썬 리스트 슬라이싱이랑 동일)
df.iloc[0:2]
```

```python
# * ::2 -> step 슬라이싱, 처음부터 끝까지 2칸씩 건너뛰기 (하나 걸러 하나)
df.iloc[::2]
```

```python
# * 대괄호 안에 값이 1개면 "행" 기준, 콤마로 나누면 [행, 열] 둘 다 지정 가능
df.iloc[0:2, 1:3]
```

```python
# * 행 자리에 :(전체) 넣고 열만 지정하면 "열만" 슬라이싱 가능
df.iloc[:, 0:2]
```

```python
# * 새 랜덤 DataFrame 다시 생성
df = pd.DataFrame(np.random.randn(4, 3), index=['a', 'b', 'c', 'd'], columns=['x', 'y', 'z'])
print(df)

# * 각 원소마다 1보다 큰지 비교 -> True/False로 이루어진 DataFrame 반환
df > 1
```

```python
# * True인 자리만 값 유지, 나머지는 NaN
df[df > 1]
```

```python
# * x열만 조건 검사 -> True/False로 이루어진 Series
df['x'] > 0
```

```python
# * x열이 0보다 큰 행만 필터링 (boolean indexing)
df[df['x'] > 0]
```

```python
# * 필터링한 행들 중에서 y열만 뽑기
df[df['x'] > 0]['y']
```

```python
df
```

```python
# * 여러 조건 합칠 땐 and/or 대신 &, | 사용 (Series라서 and/or는 안 됨)
df[(df['x'] > 0) | (df['y'] > 0)]
```

```python
# * 딕셔너리 데이터 준비
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago']
}

# * 딕셔너리로 DataFrame 생성
df = pd.DataFrame(data)

print(df)
```

```python
# * Name 열만 Series로 가져오기
df['Name']
```

```python
# * 인덱스 0번 행 가져오기
df.loc[0]
```

```python
# * Age가 30 초과인 행만 필터링
df[df['Age'] > 30]
```

```python
# * 새 열 추가 예시
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35]
}

df = pd.DataFrame(data)

# * 새 열 'City' 추가
df['City'] = ['New York', 'San Francisco', 'Los Angeles']
print(df)
```

```python
# * 열 삭제 예시
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'San Francisco', 'Los Angeles']
}

df = pd.DataFrame(data)

# * City 열 삭제 (inplace=True로 원본 자체 수정)
df.drop('City', axis=1, inplace=True)
print(df)
```
