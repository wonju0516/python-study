```python
import pandas as pd
```

```python

# * 샘플 데이터프레임(표) 만들기
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [25, 30, 35, 40, 45],
    'City': ['New York', 'San Francisco', 'Los Angeles', 'Chicago', 'Houston'],
    'Salary': [50000, 60000, 70000, 80000, 90000]
}

df = pd.DataFrame(data)  # * 딕셔너리의 key -> 열(column) 이름, value(리스트) -> 그 열의 값들

# * 데이터프레임 출력
print("Original DataFrame:")
print(df)
```

```python

# * df['Experience'] = [...] -> 없는 열 이름을 대입하면 새로운 열(column)이 추가됨 (인덱스 아님)
# * 리스트 길이는 행 개수(5개)랑 똑같아야 함
df['Experience'] = [2, 5, 8, 10, 12]
print("\nDataFrame after adding 'Experience' column:")
print(df)

# * 조건으로 행 필터링 (boolean indexing) -> Age가 30 초과인 행만 남김
older_than_30 = df[df['Age'] > 30]
print("\nPeople older than 30:")
print(older_than_30)
```

```python

# * Salary 열의 평균 계산
average_salary = df['Salary'].mean()
print(f"\nAverage Salary: {average_salary}")

# * City 별로 그룹 지어서, 각 그룹 안 숫자형 열들의 평균 계산
# * numeric_only=True 필요 -> Name처럼 문자열 열이 섞여 있으면 평균 계산 시 에러남
grouped_by_city = df.groupby('City').mean(numeric_only=True)
print("\nGrouped by City:")
print(grouped_by_city)

# * 그룹 기준을 2개 이상 쓰고 싶으면 리스트로 묶어야 함 (콤마로 나열하면 에러)
grouped_by_city_age = df.groupby(['City', 'Age']).mean(numeric_only=True)
print("\nGrouped by City and Age:")
print(grouped_by_city_age)
```
