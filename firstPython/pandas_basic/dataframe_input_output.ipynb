```python
import sqlalchemy as sa

# * 아래 값들은 예시용 자리채우개(placeholder) -> 실제 DB 접속 정보로 바꿔서 써야 함
db_username = 'user'
db_password = 'password'
db_host = '127.0.0.1'
db_port = '3306'
# ! mysqlsampledatabase.sql 안에서 실제로 만들어지는 DB 이름은 'classicmodels' -> 'db'는 존재 안 해서 접속 실패함
db_name = 'classicmodels'

# * DB 접속 문자열 조립 (mysql+드라이버://아이디:비번@호스트:포트/DB이름 형식)
connection_string = f'mysql+mysqlconnector://{db_username}:{db_password}@{db_host}:{db_port}/{db_name}'

# * create_engine -> 실제 접속은 아직 안 하고, DB랑 통신할 "연결 통로"만 준비
engine = sa.create_engine(connection_string)
```

```python
# imported the data from: https://raw.githubusercontent.com/hhorak/mysql-sample-db/master/mysqlsampledatabase.sql
```

```python
# * read_sql_query -> SQL 쿼리 결과를 바로 DataFrame으로 변환
sql_query = f'SELECT customerNumber, customerName, phone FROM customers'
data = pd.read_sql_query(sql_query, engine)

print(data.head())
```

```python
import requests
from io import StringIO

# * HTML 표가 있는 페이지 주소
url = 'https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population'

# ! pd.read_html(url)을 바로 쓰면 위키피디아가 봇 요청으로 인식해서 403 Forbidden 에러를 냄
# * headers -> 요청에 "나에 대한 정보"를 같이 실어 보내는 딕셔너리
# * User-Agent -> "나 지금 무슨 프로그램으로 접속하는 거야"를 서버에 알려주는 값
# *   기본값(안 주면)은 'python-requests/버전'이라 봇인 게 티가 남 -> 'Mozilla/5.0'으로 주면 일반 브라우저인 척 할 수 있음
headers = {'User-Agent': 'Mozilla/5.0'}

# * requests.get(url, headers=headers) -> 이 url에 이 headers(신분증)를 달고 접속해서 응답을 받아옴
response = requests.get(url, headers=headers)

# * read_html -> 페이지 안의 모든 <table>을 찾아서, 표 하나당 DataFrame 하나씩 -> DataFrame 리스트로 반환
# * StringIO로 감싸야 함 -> response.text는 그냥 문자열인데, 감싸지 않고 넘기면 "파일 경로"로 착각해서 에러남
tables = pd.read_html(StringIO(response.text))

# * 페이지에 표가 몇 개 있었는지
print(f"Number of tables found: {len(tables)}")

# * 그중 두 번째 표(인덱스 1)가 보통 원하는 데이터인 경우가 많음
print(tables[1].head())
```
